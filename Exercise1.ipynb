> Import the required libraries for data processing, visualization, model training, and evaluation.

```
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
)

# ============================================================
#  🎨 Shared visual theme — used consistently across all plots
# ============================================================
PALETTE = {
    "introvert": "#6C5CE7",   # violet
    "extrovert": "#00B894",   # teal green
    "accent":    "#2D3436",   # near-black text
    "muted":     "#636E72",   # secondary grey text
    "grid":      "#DFE6E9",   # soft gridlines
    "bg":        "#FAFAFC",   # panel background
}
CMAP_VIOLET = plt.cm.colors.LinearSegmentedColormap.from_list(
    "violet_teal", ["#6C5CE7", "#00B894"]
)

plt.rcParams.update({
    "figure.facecolor": "white",
    "axes.facecolor": PALETTE["bg"],
    "axes.edgecolor": PALETTE["grid"],
    "axes.labelcolor": PALETTE["accent"],
    "text.color": PALETTE["accent"],
    "xtick.color": PALETTE["muted"],
    "ytick.color": PALETTE["muted"],
    "axes.titleweight": "bold",
    "axes.titlesize": 14,
    "axes.titlecolor": PALETTE["accent"],
    "font.size": 10.5,
    "font.family": "sans-serif",
    "axes.spines.top": False,
    "axes.spines.right": False,
    "axes.spines.left": False,
    "axes.spines.bottom": False,
    "axes.grid": True,
    "grid.color": PALETTE["grid"],
    "grid.linewidth": 0.7,
})
```

### Display the total number of samples from each label using `.value_counts()`

```
personality_df["Personality"].value_counts()
```

```
# --- Visualize the class balance as a donut chart ---
counts = personality_df["Personality"].value_counts()
labels = ["Introvert" if l == 0 else "Extrovert" for l in counts.index] if counts.index.dtype != object else list(counts.index)
colors = [PALETTE["introvert"] if str(l).lower().startswith("intro") else PALETTE["extrovert"] for l in labels]

fig, ax = plt.subplots(figsize=(5.5, 5.5))
wedges, _, autotexts = ax.pie(
    counts.values,
    colors=colors,
    startangle=90,
    autopct=lambda p: f"{p:.1f}%",
    pctdistance=0.8,
    wedgeprops=dict(width=0.42, edgecolor="white", linewidth=3),
    textprops=dict(color="white", fontsize=11, fontweight="bold"),
)
ax.text(0, 0.05, f"{counts.sum()}", ha="center", va="center", fontsize=22, fontweight="bold", color=PALETTE["accent"])
ax.text(0, -0.15, "samples", ha="center", va="center", fontsize=11, color=PALETTE["muted"])
ax.legend(labels, loc="upper center", bbox_to_anchor=(0.5, -0.02), ncol=2, frameon=False)
ax.set_title("Class Distribution: Introvert vs Extrovert", pad=18)
ax.set(aspect="equal")
plt.tight_layout()
plt.show()
```

### Use a histogram to display the number of followers/friends for each sample

```
fig, ax = plt.subplots(figsize=(8.5, 4.5))
counts, bins, patches = ax.hist(
    personality_df["Friends_circle_size"],
    bins=12,
    edgecolor="white",
    linewidth=1.2,
)

# Color each bar along a violet -> teal gradient based on its height
norm = plt.Normalize(counts.min(), counts.max())
for count, patch in zip(counts, patches):
    patch.set_facecolor(CMAP_VIOLET(norm(count)))

mean_val = personality_df["Friends_circle_size"].mean()
ax.axvline(mean_val, color=PALETTE["accent"], linestyle="--", linewidth=1.3, alpha=0.7)
ax.text(mean_val, ax.get_ylim()[1]*0.95, f"  mean = {mean_val:.1f}", color=PALETTE["accent"], fontsize=9.5, va="top")

ax.set_title("Friends Circle Size Distribution")
ax.set_xlabel("Friends Circle Size")
ax.set_ylabel("Number of Samples")
plt.tight_layout()
plt.show()
```

### Convert the features into a NumPy array

```
personality_df["Stage_fear"]=personality_df["Stage_fear"].replace({"Yes":1,"No":0})
personality_df["Drained_after_socializing"]=personality_df["Drained_after_socializing"].replace({"Yes":1,"No":0})
personality_df["Personality"]=personality_df["Personality"].replace({"Introvert":0,"Extrovert":1})

X = personality_df.iloc[:, :-1].to_numpy()
```

### Convert the labels into a NumPy array

```
y = personality_df.iloc[:, -1].to_numpy()
```

### Split the dataset

```
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    shuffle=True
)
```

### Display the dimensions

```
print("X_train:",X_train.shape)
print("X_test :",X_test.shape)
print("y_train:",y_train.shape)
print("y_test :",y_test.shape)
```

<div style="background: linear-gradient(90deg, #E17055 0%, #fab1a0 100%); padding: 14px 22px; border-radius: 10px; color: white; font-family: 'Segoe UI', sans-serif; display:flex; align-items:center; gap:12px; box-shadow: 0 3px 10px rgba(0,0,0,0.12);">
<div style="font-size:22px;">🤖</div>
<div>
<div style="font-size:11px; opacity:0.8; letter-spacing:2px;">SECTION D</div>
<div style="font-size:18px; font-weight:700;">Create the Model</div>
</div>
</div>

```
knn_model = KNeighborsClassifier(n_neighbors=5)

knn_model.fit(X_train, y_train)

predictions = knn_model.predict(X_test)
```

<div style="background: linear-gradient(90deg, #00B894 0%, #55efc4 100%); padding: 14px 22px; border-radius: 10px; color: white; font-family: 'Segoe UI', sans-serif; display:flex; align-items:center; gap:12px; box-shadow: 0 3px 10px rgba(0,0,0,0.12);">
<div style="font-size:22px;">📈</div>
<div>
<div style="font-size:11px; opacity:0.8; letter-spacing:2px;">SECTION E</div>
<div style="font-size:18px; font-weight:700;">Evaluate the Model</div>
</div>
</div>

```
cm = confusion_matrix(y_test, predictions)
acc = accuracy_score(y_test, predictions)
prec = precision_score(y_test, predictions)
rec = recall_score(y_test, predictions)
f1 = f1_score(y_test, predictions)

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), gridspec_kw={"width_ratios": [1, 1.3]})

# --- Left: confusion matrix ---
ax = axes[0]
im = ax.imshow(cm, cmap=CMAP_VIOLET, alpha=0.9)
ax.set_xticks([0, 1]); ax.set_xticklabels(["Introvert", "Extrovert"])
ax.set_yticks([0, 1]); ax.set_yticklabels(["Introvert", "Extrovert"])
ax.set_xlabel("Predicted label")
ax.set_ylabel("True label")
ax.set_title("Confusion Matrix")
ax.grid(False)
thresh = cm.max() / 2
for r in range(cm.shape[0]):
    for c in range(cm.shape[1]):
        color = "white" if cm[r, c] > thresh else PALETTE["accent"]
        ax.text(c, r, str(cm[r, c]), ha="center", va="center", fontsize=15, fontweight="bold", color=color)

# --- Right: metric bars ---
ax2 = axes[1]
metrics = ["Accuracy", "Precision", "Recall", "F1 Score"]
scores = [acc, prec, rec, f1]
bar_colors = [CMAP_VIOLET(x) for x in np.linspace(0.15, 0.9, len(metrics))]
bars = ax2.barh(metrics, scores, color=bar_colors, height=0.55)
ax2.set_xlim(0, 1.08)
ax2.invert_yaxis()
ax2.set_title("Model Performance Summary")
ax2.grid(axis="x", alpha=0.4)
for bar, score in zip(bars, scores):
    ax2.text(score + 0.02, bar.get_y() + bar.get_height()/2, f"{score:.3f}",
             va="center", fontsize=11, fontweight="bold", color=PALETTE["accent"])

plt.tight_layout()
plt.show()

print(f"Accuracy : {acc:.3f}")
print(f"Precision: {prec:.3f}")
print(f"Recall   : {rec:.3f}")
print(f"F1 Score : {f1:.3f}")
```
